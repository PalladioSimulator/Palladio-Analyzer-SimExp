```python
import pandas as pd
from pathlib import Path
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
strategy = "2d"
folder = "DeltaIoT_modelled_ea_strategy%s" % strategy
filename = "%s_cumulative_pareto_front_rank.csv" % folder
filename_pdf = "%s_cumulative_pareto_front_rank.pdf" % folder
filename_png = "%s_cumulative_pareto_front_rank.png" % folder
```

```python
simexp_path = Path.cwd().parent / "simulation_result"
```

```python
#df = pd.read_csv(simexp_path / "resources" / "DeltaIoT_modelled_ea_strategy0a_cumulative_pareto_front_rank.csv")
df = pd.read_csv(simexp_path / "resources" / filename)
```

```python
df.head()
```

```python
sns.lineplot(data=df, x="generation", y="hv", marker="o")

image_folder = simexp_path / "resources" / folder / "figures"
image_folder.mkdir(exist_ok=True)
plt.savefig(image_folder / filename_png, dpi=300, bbox_inches="tight")
plt.savefig(image_folder / filename_pdf, dpi=300, bbox_inches="tight")

plt.show()
```
